```python
# 06.09.2026 - 20260906m.ipynb: RYSUNKI DO PREZENTACJI, PO ANGIELSKU.
# WERSJA 2 TEGO NOTEBOOKA = deck w wersji 3, po uwagach Macka
# (prezentacja/mysli_v2.txt, wieczor 06.09). Numeracja notebooka i decku sie
# rozjezdza, bo deck ma za soba jeszcze runde po mysli_v1.txt.
#
# JEDEN watek: wyprodukowac komplet rysunkow decku. Notebook prawie NICZEGO
# NIE BADA - czyta zapisane wyniki i rysuje. Wyjatek jest jeden i jawny:
# komorka 1 liczy od zera krzywa sekcji 4 na pliku CR Z ARCHIWUM ZESPOLU
# (18 s), bo tylko na tym pliku porownanie z opublikowanymi glebokosciami
# jest jak za jak - patrz komentarz tam.
#
# DLACZEGO WERSJA 2 - CO BYLO ZLE I JAK TO NAPRAWIONO.
# Uwaga Macka byla jednoznaczna: "wykresy wygladaja horendalnie, skala jest za
# mala, caly tekst sie zlewa i nakłada na siebie". Przyczyna byla JEDNA i
# opisuje ja tu w calosci, zeby nie wrocila.
#
#   BLAD. Wersja 1 miala stala SKALA = 0,70, ktora zmniejszala PLOTNO rysunku
#   (figsize), zostawiajac czcionki na 9,5 pt. Arytmetyka byla poprawna
#   (mniejszy plik na tej samej szerokosci slajdu = wieksza czcionka na
#   ekranie), ale skutek uboczny zabil rysunki: przy plotnie wysokim na
#   1,7-2,6 cala dwuwierszowy tytul, podpisy osi i adnotacje po prostu nie
#   miesza sie obok siebie i NACHODZA NA SIEBIE. Na rysunku korelacji dwa
#   podpisy osi Y lezaly jeden na drugim; na rysunku Poissona adnotacja
#   "pure Poisson noise" lezala na slupkach; na rysunku kruchosci podpisy
#   punktow lezaly na tytule.
#
#   ZASADA POPRAWKI (Maciek, 06.09): "rob wykresy jakby mialy byc normalnymi
#   wykresami do pokazania; to zupelnie normalne, ze wykres ma mniejsza
#   czcionke niz prezentacja, nie probuj tego zrownac".
#
#   CO Z NIEJ WYNIKA, PUNKT PO PUNKCIE:
#    0. UWAGA, STAN NA 06.09 WIECZOREM: figsize kazdego rysunku zostal
#       PRZEMNOZONY PRZEZ 2,5 PRZEZ MACKA, po obejrzeniu wynikow. Zapis
#       "2.5*4.00" jest celowo zostawiony w tej postaci - to jego mnoznik i
#       jego decyzja ("nie ruszaj juz tego, jak cos to ja to bede zmienial").
#       Jedyne, co ruszamy nizej, to SKLADOWA X dwoch rysunkow (motivation,
#       instrument), o ktora prosil osobno, zeby panele nie byly kwadratowe.
#       KONSEKWENCJA, KTORA TRZEBA ZNAC: czcionki zostaly na 10 pt, wiec
#       tekst na rysunku jest teraz 2,5x mniejszy WZGLEDEM plotna - pliki
#       PNG wygladaja czysto, ale na planszy renderuja sie jako ~3,9 pt.
#       Zeby to zmienic, nie rusza sie figsize, tylko mnozy czcionki w
#       rcParams i w hardkodowanych fontsize= przez ten sam czynnik.
#       Akapity 1-4 nizej opisuja reguly, wedlug ktorych rozmiary byly
#       dobrane wczesniej - zostaja jako zapis rozumowania.
#
#    1. PROPORCJE JAK W NORMALNYM MALYM WYKRESIE. Szerokosc 4,0 cala,
#       wysokosc 1,6-2,4 cala, czyli stosunek 1,7-2,4:1. Nigdy 8,2 x 1,7.
#       WASKIE PLOTNO JEST TU CELOWE I TO JEST KONTRINTUICYJNE: czcionka
#       widziana na planszy rosnie odwrotnie proporcjonalnie do szerokosci
#       PLIKU, wiec 4,0 cala przy 10 pt daje na planszy WIECEJ niz 6,8 cala
#       przy 10 pt. Wersja 1 miala to samo rozumowanie i dlatego nie byla
#       glupia - wykonala je zle, sciskajac tylko WYSOKOSC i zostawiajac na
#       plotnie tyle samo elementow. Tutaj plotno jest waskie ROWNOMIERNIE i
#       ma na sobie MNIEJ elementow.
#    2. ZERO TYTULOW WEWNATRZ RYSUNKU. Komunikat niesie tytul planszy. Tytuly
#       wewnetrzne byly zrodlem polowy kolizji (dwa wiersze nad ciasnym
#       plotnem) i dublowaly tekst, ktory i tak stoi na planszy. Krotkie,
#       jednowierszowe podpisy PANELI zostaja - to etykieta, nie komunikat.
#    3. ARYTMETYKA DOCELOWEJ CZCIONKI. Czcionka o rozmiarze F w pliku
#       szerokim na S cali renderuje sie na planszy jako F * W / (S * 72,27)
#       punktow, gdzie W to szerokosc, jaka rysunek dostaje w LaTeX-u.
#       Szerokosc tekstu planszy to 398 pt. Przy S = 4,0, F = 10 i
#       W = 0,60 * 398 = 239 pt wychodzi 8,3 pt. Zmierzone dla calego decku:
#       poprzednia wersja tego notebooka dawala 3,5-7,3 pt (mediana 5,5),
#       ta daje 6,9-8,5 pt
#       (mediana 8,1). Rysunki wchodza na plansze na 0,50-0,62 \linewidth i
#       stoja NA MAKSIMUM, jakie dopuszcza tekst planszy - sprawdzone
#       automatem (rozszerzanie co 0,02 az do pierwszego przepelnienia).
#       Zeby czcionka urosla wyzej, trzeba zabrac planszy tekst.
#    4. MNIEJ ELEMENTOW. Jesli po tych trzech punktach cos dalej nie miesci
#       sie na rysunku, to znaczy, ze rysunek ma za duzo tresci - wyrzucamy
#       tresc, nie zmniejszamy czcionke.
#
# ZASADY RYSUNKOWE (bez zmian z wersji 1, nadal obowiazuja):
#  1. ZADNEJ podwojnej osi Y. Dwie wielkosci o roznej skali -> dwa panele.
#     JEDEN WYJATEK, swiadomy: rysunek 1 (plansza 2). Ma byc rozpoznawalny
#     jako Fig. 4 artykulu, a Fig. 4 ma plamy sloneczne na drugiej osi.
#     Decyzja Macka, 06.09: "wykres calkowicie w stylu Fig. 4, tzn. na jednym
#     wykresie obie krzywe". Nigdzie indziej tego nie robimy.
#  2. Jedno kodowanie na wykres (zasada Macka z 26.08): polozenie ALBO
#     wysokosc ALBO kolor - nigdy trzy naraz.
#  3. Kolory z ustalonej palety, w ustalonej kolejnosci, nigdy cyklicznie.
#     Rola stala w calym decku: POMARANCZOWY = to, z czym sie porownujemy
#     (wartosc opublikowana albo katalog prawdziwy), NIEBIESKI = nasze /
#     konstrukcja testowa, SZARY = kontekst i progi.
#  4. Serie podpisane bezposrednio przy krzywej, gdy jest ich <= 4.
#  5. Kierunek osi: wszedzie -log10(PPDF), rosnaco = silniejszy efekt.
#     UWAGA: to jest ODWROTNIE niz na Fig. 4 artykulu, ktora rysuje
#     log10(PPDF) i minima w dol. Wybor swiadomy: wszystkie liczby, ktore
#     cytujemy na planszach (9,72; 7,84; 6,47 sigma) sa dodatnie, wiec os
#     rosnaca zgadza sie z tekstem, a widz nie musi odwracac znaku w glowie.
#     Ksztalt rysunku i tak jest rozpoznawalny - te same dane, ta sama os
#     czasu, te same zaznaczone ery. Dlatego na planszach mowimy "eras",
#     a nie "minima".
#
# WEJSCIA (wszystko na dysku):
#   ../źródła/kod/dane/mosc_data2.csv        (plik CR zespolu, 1960-2019)
#   ../data/usgs_data/usgs_m4_1965_2025.csv
#   ../data/mosc_data.csv, ../data/SN_m_tot_V2.0.csv
#   ../results/credo_czynniki_fig4_krzywe_mosc.csv     (58 MB, plan 2x2)
#   ../results/credo_mosc_P3350_d5_dt15_fullhistory.csv
#   ../results/credo_kotwice.csv
#   ../results/kalibracja_surogatowa_lancuch.csv (+ _podsumowanie)
#   ../results/local_median_robustness_summary.csv
# WYJSCIA: ../results/deck_*_en.png  (dziesiec rysunkow)
import os
import re
import sys
import importlib

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

sys.path.insert(0, "..")
import statystyka as st
importlib.reload(st)

RESULTS_DIR = "../results"
DATA_DIR = "../data"
ARCH_DIR = "../źródła/kod/dane/"
DPI = 200

# --- paleta (slot 1-3 walidowanej palety kategorycznej + szarosci) ----------
NIEBIESKI = "#2a78d6"   # nasze / konstrukcja testowa
POMARANCZ = "#eb6834"   # to, z czym sie porownujemy
AQUA      = "#1baf7a"   # trzecia seria, gdy naprawde potrzebna
TEKST     = "#0b0b0b"
TEKST2    = "#52514e"
SZARY     = "#9a9994"
SIATKA    = "#e3e2de"

# JEDNA POKRETLO NA WIELKOSC LITER NA RYSUNKACH.
#
# PO CO. Figsize kazdego rysunku jest przemnozony przez 2,5 (decyzja Macka,
# 06.09, po obejrzeniu wynikow - przy 1,0 podpisy na siebie wchodzily).
# Czcionki zostaly przy tym na 10 pt, wiec tekst zajmuje teraz 2,5 raza
# mniejsza czesc plotna: pliki PNG wygladaja czysto, ale NA PLANSZY renderuja
# sie jako okolo 3,9 pt, bo wielkosc liter na slajdzie to
#     F * W / (S * 72,27),
# gdzie F - czcionka w pliku, S - szerokosc PLIKU w calach, W - szerokosc,
# jaka rysunek dostaje w LaTeX-u (tu 240-400 pt).
#
# JAK TEGO UZYWAC. Nie ruszaj figsize - zmien ponizsza liczbe i odpal
# notebook ponownie. Wszystkie czcionki (rcParams i te wpisane wprost w
# wywolaniach) sa przez nia przemnozone, wiec proporcje rysunku zmieniaja sie
# spojnie i nic sie nie rozjedzie.
#   1,0  - stan obecny: czysto w pliku PNG, ~3,9 pt na planszy
#   1,6  - bezpieczny srodek: ~6,2 pt na planszy, tekst wciaz zajmuje mniej
#          plotna niz w wersji, ktora wchodzila na siebie
#   2,5  - dokladnie proporcje sprzed przemnozenia figsize: ~9,8 pt na
#          planszy, ale to jest ta wersja, w ktorej podpisy kolidowaly
# Przy zmianie warto potem przejrzec rysunki, czy nic na siebie nie wchodzi.
CZCIONKA = 1.0

mpl.rcParams.update({
    # savefig.bbox CELOWO NIE "tight". Przy "tight" matplotlib PRZYCINA
    # zapisany obrazek do zawartosci, wiec proporcje pliku zaleza od tego, ile
    # miejsca zajely akurat podpisy - i roznia sie od figsize nawet o 15%.
    # Deck dobiera szerokosci rysunkow na planszy pod ZNANE proporcje (patrz
    # prezentacja/README.md), wiec plik musi miec dokladnie te, ktore tu
    # stoja. tight_layout() nizej i tak dopasowuje uklad WEWNATRZ plotna,
    # wiec nic sie nie obcina - pojawia sie tylko biala ramka, niewidoczna na
    # bialym tle planszy.
    "figure.dpi": DPI, "savefig.dpi": DPI, "savefig.bbox": "standard",
    "font.family": "DejaVu Sans", "font.size": 10.0 * CZCIONKA,
    "axes.titlesize": 10.0 * CZCIONKA, "axes.labelsize": 10.0 * CZCIONKA,
    "axes.labelcolor": TEKST2, "axes.edgecolor": SZARY, "axes.linewidth": 0.9,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": SIATKA, "grid.linewidth": 0.9,
    "xtick.color": TEKST2, "ytick.color": TEKST2,
    "xtick.labelsize": 9.0 * CZCIONKA, "ytick.labelsize": 9.0 * CZCIONKA,
    "legend.frameon": False, "legend.fontsize": 9.0 * CZCIONKA,
    "figure.facecolor": "white", "axes.facecolor": "white",
})


ROZMIARY = {}          # nazwa -> (szerokosc, wysokosc) w calach; uzywa tego
                       # diagnostyka czcionki w ostatniej komorce


def zapisz(fig, nazwa):
    sciezka = f"{RESULTS_DIR}/deck_{nazwa}_en.png"
    fig.savefig(sciezka)
    w, h = fig.get_size_inches()
    plt.close(fig)
    ROZMIARY[f"deck_{nazwa}"] = (float(w), float(h))
    print(f"  zapisano {sciezka}  ({w:.2f} x {h:.2f} cala)")
    return sciezka


# Formatowanie dat RECZNIE: mpl.dates.DateFormatter idzie przez locale
# systemu, a ten jest polski - na rysunku po angielsku wychodzilo "08 lut".
MIES = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct",
        "Nov", "Dec"]


def po_ang(t):
    return f"{t.day} {MIES[t.month - 1]} {t.year}"


ZAPISANE = []

# --- dane wspolne ----------------------------------------------------------
P_DAYS, D_DAYS, DT_DAYS, M_THR = 3350, 5, 15, 4.0
OKNO = st.okno(P_DAYS, D_DAYS, DT_DAYS, st.CREDO)

eq = pd.read_csv(f"{DATA_DIR}/usgs_data/usgs_m4_1965_2025.csv", usecols=["time", "mag"])
eq["time"] = pd.to_datetime(eq["time"], utc=True).dt.tz_localize(None)
eq = eq[eq["mag"] >= M_THR].sort_values("time").reset_index(drop=True)
eq["rok"] = eq["time"].dt.year
# pelne lata katalogu (1965 zaczyna sie 01-01, 2025 jest ucieta)
LATA = np.arange(1965, 2025)


def licznik(m, lata=None):
    lata = LATA if lata is None else lata
    return (eq[eq["mag"] >= m].groupby("rok").size()
            .reindex(lata, fill_value=0).astype(float))


# plamy sloneczne - potrzebne na rysunku 1 (dolny panel)
sn = pd.read_csv(f"{DATA_DIR}/SN_m_tot_V2.0.csv", sep=";")
sn = sn[sn["sunspot"] >= 0].copy()
sn["data"] = pd.to_datetime(dict(year=sn["rok"], month=sn["miesiac"], day=15))
sn["gladka"] = sn["sunspot"].rolling(13, center=True, min_periods=7).mean()

print(f"Katalog: {len(eq)} zdarzen M >= {M_THR}, lata pelne {LATA[0]}-{LATA[-1]}")
print(f"Plamy sloneczne: {len(sn)} miesiecy, {sn['data'].min():%Y-%m} .. {sn['data'].max():%Y-%m}")
print(f"Paleta: nasze {NIEBIESKI}, odniesienie {POMARANCZ}, trzecia {AQUA}")
```

```python
# RYSUNEK 1 (plansza 2) - NASZA KRZYWA SEKCJI 4 W UKLADZIE FIG. 4 ARTYKULU.
#
# CO ZASTEPUJE I DLACZEGO. Wersja 1 miala tu hantel: piec opublikowanych
# glebokosci kontra nasze, w jednym rzedzie na osi poziomej. Maciek: "tu ma
# zostac wykres w stylu Fig. 4, nie po to zrobiles go bardzo ladnego, zeby go
# teraz nie uzyc". Ma racje takze merytorycznie - hantel pokazuje PIEC LICZB,
# a Fig. 4 pokazuje, ze te piec liczb siedzi w rytmie widocznym golym okiem na
# calej pieciodziesiecioletniej krzywej. To jest mocniejsze stwierdzenie.
#
# DLACZEGO LICZYMY KRZYWA OD ZERA, A NIE CZYTAMY GOTOWEJ.
# Gotowa krzywa (credo_mosc_P3350_d5_dt15_fullhistory.csv) jest policzona na
# NASZYM pobraniu Moskwy z NMDB. Opublikowane glebokosci odtwarzamy z
# dokladnoscia 0,24-0,77 dekady (suma 1,93) na pliku CR Z ARCHIWUM ZESPOLU, a
# nie na naszym - sprawdzone w 20260906a. Na naszym pliku ta sama piatka dat
# wypada plycej. Rysunek, ktory pokazuje NASZA krzywa i OPUBLIKOWANE punkty,
# musi wiec byc policzony na pliku zespolu, inaczej znaczniki wisialyby nad
# krzywa i obrazek mowilby cos, czego slajd nie mowi.
#
# Koszt: siatka co 6 h przez cala historie pliku archiwalnego to okolo 66 tys.
# punktow po 0,2 ms, czyli okolo 15 sekund. Podpis pod rysunkiem na planszy
# MUSI mowic, ze to plik CR zespolu.
#
# FAZA SIATKI. Opublikowane t0 stoja o 07:37:12. Krzywa oscyluje o kilka
# dekad z okresem rownym szerokosci kosza (patrz deck_fragility), wiec siatka
# co 6 h ZAKOTWICZONA GDZIE INDZIEJ minelaby te punkty i krzywa wygladalaby
# plycej, niz jest. Dlatego siatke kotwiczymy w fazie artykulu: 01:37:12 + k*6h
# przechodzi dokladnie przez 07:37:12.
ARTYKUL = pd.DataFrame([
    dict(nr=1, t0=pd.Timestamp("1968-07-12 07:37:12"), PPDF=3.34e-8),
    dict(nr=2, t0=pd.Timestamp("1978-07-11 07:37:12"), PPDF=3.00e-6),
    dict(nr=3, t0=pd.Timestamp("1988-09-21 07:37:12"), PPDF=3.58e-5),
    dict(nr=4, t0=pd.Timestamp("1999-12-23 07:37:12"), PPDF=2.92e-6),
    dict(nr=5, t0=pd.Timestamp("2009-02-23 07:37:12"), PPDF=8.10e-9),
])
ARTYKUL["opublikowane"] = -np.log10(ARTYKUL["PPDF"])

PROG_ERY = 5.0
POL_OKNA = pd.Timedelta(days=P_DAYS / 2)


def load_cr(path):
    df = pd.read_csv(path)
    df["datetime"] = pd.to_datetime(df["datetime"])
    s = df.set_index("datetime").sort_index()["value"]
    return s[~s.index.duplicated(keep="first")]


mosc_arch = load_cr(ARCH_DIR + "mosc_data2.csv")
crp = st.przygotuj_szereg(mosc_arch)
eqp = st.przygotuj_katalog(eq.set_index("time")["mag"])

FAZA = pd.Timedelta(hours=1, minutes=37, seconds=12)     # 07:37:12 mod 6 h
start = max(pd.Timestamp("1965-01-01"), mosc_arch.index[0]).normalize() + FAZA
koniec = mosc_arch.index[-1] - pd.Timedelta(days=P_DAYS)
siatka = pd.date_range(start, koniec, freq="6h")
assert all(t in set(siatka) for t in ARTYKUL["t0"]), "siatka minela punkty artykulu"

y, _Np, _Nm, _Nn, _dt = st.krzywa(
    crp, eqp, siatka.values.astype("datetime64[ns]").astype(np.int64),
    OKNO, max_jitter_d=D_DAYS)
kr = pd.DataFrame(dict(t0=siatka, y=y))
kr["srodek"] = kr["t0"] + POL_OKNA
print(f"Krzywa na pliku zespolu: {len(kr)} punktow co 6 h, "
      f"{kr['t0'].iloc[0]:%Y-%m-%d} .. {kr['t0'].iloc[-1]:%Y-%m-%d}; "
      f"odrzuconych progiem higieny {st.krzywa.ostatnio_odrzuconych}")

# nasze ery: find_peaks z separacja rowna dlugosci okna (patrz notatki - to
# jest tez powod, dla ktorego NIE WOLNO czytac odstepow miedzy erami jako
# potwierdzenia rytmu ~11-letniego)
_y = np.nan_to_num(kr["y"].values, nan=-1.0)
ery_i, _ = find_peaks(_y, height=PROG_ERY, distance=int(P_DAYS / 0.25))
nasze = kr.iloc[ery_i]
print(f"Nasze ery ({len(nasze)}): " + ", ".join(
    f"{t:%Y-%m-%d} ({v:.2f})" for t, v in zip(nasze["t0"], nasze["y"])))

w_punktach = [float(kr.loc[(kr["t0"] - t).abs().idxmin(), "y"]) for t in ARTYKUL["t0"]]
ARTYKUL["nasze"] = w_punktach
ARTYKUL["roznica"] = ARTYKUL["nasze"] - ARTYKUL["opublikowane"]

# KONTROLA, KTORA UZASADNIA WYBOR PLIKU. Te same piec dat, policzone na NASZYM
# pobraniu Moskwy z NMDB. Jesli roznica jest tam wyraznie wieksza, to znaczy,
# ze rysunek MUSI stac na pliku zespolu - i mamy to czarno na bialym, zamiast
# powolywac sie na pamiec. Piec punktow, wiec kosztuje ulamek sekundy.
mosc_nasz = load_cr(f"{DATA_DIR}/mosc_data.csv")
crp_nasz = st.przygotuj_szereg(mosc_nasz)
ARTYKUL["nasz_plik"] = [-np.log10(st.punkt(crp_nasz, eqp, t, OKNO)["PPDF"])
                        for t in ARTYKUL["t0"]]
ARTYKUL["roznica_nasz"] = ARTYKUL["nasz_plik"] - ARTYKUL["opublikowane"]

print(ARTYKUL[["nr", "opublikowane", "nasze", "roznica", "nasz_plik", "roznica_nasz"]]
      .to_string(index=False, float_format=lambda x: f"{x:6.2f}"))
print(f"plik zespolu: suma |roznicy| = {ARTYKUL['roznica'].abs().sum():.2f} dekady, "
      f"najgorsza era {ARTYKUL['roznica'].abs().max():.2f}")
print(f"nasze pobranie: suma |roznicy| = {ARTYKUL['roznica_nasz'].abs().sum():.2f}, "
      f"najgorsza {ARTYKUL['roznica_nasz'].abs().max():.2f}  "
      f"(NaN = punkt odrzucony progiem higieny)")

# UKLAD: JEDNA OS, OBIE KRZYWE, log10(PPDF) - jak Fig. 4 artykulu.
# Decyzja Macka, 06.09. To jedyne miejsce w decku z podwojna osia Y i jedyne
# z osia log10 malejaca w dol; obie rzeczy sa tu po to, zeby rysunek byl
# rozpoznawalny jako TA SAMA figura, co w artykule. Konsekwencja dla tekstu:
# na planszy mowimy "eras", a glebokosci cytujemy jako -log10, wiec plansza i
# os maja przeciwne znaki - dlatego os jest podpisana wprost, a nie skrotem.
fig, ax = plt.subplots(figsize=(2.5*4.00, 2.5*1.91))

# plamy sloneczne NAJPIERW i pod spodem - to tlo, nie wynik
ax2 = ax.twinx()
ax2.plot(sn["data"], sn["gladka"], color=SZARY, lw=1.8, zorder=1)
ax2.set_ylabel("sunspot number", color=SZARY)
ax2.tick_params(axis="y", colors=SZARY)
ax2.spines["right"].set_visible(True)
ax2.spines["right"].set_color(SZARY)
ax2.set_ylim(0, 260)
ax2.grid(False)

ax.set_zorder(ax2.get_zorder() + 1)     # krzywa PPDF nad plamami
ax.patch.set_visible(False)
ax.plot(kr["srodek"], -kr["y"], color=NIEBIESKI, lw=0.45, zorder=2)
ax.axhline(-PROG_ERY, color=SZARY, lw=1.0, ls=(0, (4, 3)), zorder=1)
ax.scatter(ARTYKUL["t0"] + POL_OKNA, -ARTYKUL["opublikowane"], marker="v",
           s=95, color=POMARANCZ, zorder=4, edgecolor="white", linewidth=0.8)
ax.annotate("published in the paper", (ARTYKUL["t0"].iloc[0] + POL_OKNA,
                                       -ARTYKUL["opublikowane"].iloc[0]),
            xytext=(9, -3), textcoords="offset points", color=POMARANCZ,
            fontsize=10 * CZCIONKA, fontweight="bold", va="top")
ax.annotate("our scan, station MOSC", (kr["srodek"].iloc[0], -15.8),
            xytext=(4, 0), textcoords="offset points", color=NIEBIESKI,
            fontsize=10 * CZCIONKA, fontweight="bold", va="bottom")
ax.set_ylabel(r"$\log_{10}\,P_{\mathrm{PDF}}$   (deeper = stronger)")
ax.set_ylim(-16.5, 0)
ax.set_xlabel("centre of the 3350-day analysis window")
ax.set_xlim(kr["srodek"].iloc[0] - pd.Timedelta(days=200),
            kr["srodek"].iloc[-1] + pd.Timedelta(days=200))
fig.tight_layout()
ZAPISANE.append(zapisz(fig, "replication"))
```

```python
# RYSUNEK 5 (plansza 6, "Step 2") - WIELOLETNI KSZTALT KATALOGU JEST
# PRZYRZADEM, NIE ZIEMIA.
#
# CO ZASTEPUJE. Wersja 1 miala tu DWA osobne rysunki na dwoch planszach:
# roczny licznik przy trzech progach (plansza 3) i nadwyzka nad szumem
# Poissona (plansza 5). Maciek zglosil dwie rzeczy: (a) "w punkcie drugim
# zostawiles wytlumaczenie b, ale wykres sie juz do niego nie odnosi -
# podoba mi sie to wytlumaczenie, zmien wykres", (b) surowy wzrost M4 ma byc
# pokazany "najprostsza wersja". Rozwiazanie: surowy wzrost przeniosl sie na
# plansze 3 (motywacja), a TUTAJ stoja dwa POMIARY, ktore na pytanie z
# planszy 3 odpowiadaja - i lewy panel jest teraz dokladnie tym, o czym mowi
# punkt o b.
#
# LEWY PANEL - IMPLIKOWANY WYKLADNIK b. Gutenberg-Richter: log10 N(>=M) =
# a - bM. Majac liczniki przy dwoch progach M1 < M2 mozna b odzyskac:
# b = log10( N(>=M1) / N(>=M2) ) / (M2 - M1). Jesli katalog jest kompletny
# powyzej M1, ta liczba MUSI byc stala w czasie. Rysujemy ja po dekadach dla
# dwoch par progow.
#
# PRAWY PANEL - ZMIENNOSC ROCZNIKA W JEDNOSTKACH PODLOGI POISSONA.
# Co dokladnie liczymy, bo nazwa "scatter" okazala sie nieczytelna (uwaga
# Macka, 06.09) i zostala zamieniona na "year-to-year variation":
#   1. bierzemy roczna liczbe zdarzen powyzej progu, N(rok);
#   2. odejmujemy od niej gladki trend (wielomian 3. stopnia) - zostaje samo
#      wahanie z roku na rok, bez wieloletniego dryfu;
#   3. liczymy odchylenie standardowe tego wahania: o ile typowo rocznik
#      odbiega od swojego trendu;
#   4. dzielimy je przez sqrt(sredniego N).
# Dlaczego akurat przez sqrt(N): jesli trzesienia byly niezalezne (proces
# Poissona), to liczba zdarzen w roku ma z definicji odchylenie standardowe
# rowne sqrt(N) - to jest NIEUSUWALNA podloga, biorąca sie z samego liczenia,
# a nie z zadnej fizyki. Iloraz mowi wiec: "obserwowane wahanie jest X razy
# wieksze niz to, ktore musialoby byc nawet przy calkowicie niezaleznych
# trzesieniach". 1x = dokladnie tyle, ile daje samo liczenie.
# PO CO NAM TO: nadwyzke ponad 1x moze tlumaczyc prawdziwe grupowanie
# sejsmiczne (wstrzasy wtorne). Ale sekwencja wstrzasow wtornych zawiera
# zdarzenia KAZDEJ wielkosci, wiec gdyby to bylo wyjasnienie, nadwyzka
# musialaby byc widoczna takze przy duzych magnitudach. Nie jest: spada
# monotonicznie z progiem az do 1,3x przy M >= 6,5.
PARY = [(4.0, 5.0), (5.5, 6.5)]
DEK = [(1965, 1974), (1975, 1984), (1985, 1994), (1995, 2004), (2005, 2014),
       (2015, 2024)]
PROGI_P = [4.0, 4.5, 5.0, 5.5, 6.0, 6.5]
POLOWKI = {"1965-1994": (1965, 1994), "1995-2024": (1995, 2024)}


def implikowane_b(m1, m2, od, do):
    w = eq[(eq["rok"] >= od) & (eq["rok"] <= do)]
    n1 = (w["mag"] >= m1).sum()
    n2 = (w["mag"] >= m2).sum()
    return float(np.log10(n1 / n2) / (m2 - m1))


def nadwyzka(m, od, do, stopien=3):
    n = licznik(m)
    trend = np.polyval(np.polyfit(LATA, n.values, stopien), LATA)   # wspolny trend
    reszta = n.values - trend
    maska = (LATA >= od) & (LATA <= do)
    return float(np.std(reszta[maska], ddof=1) / np.sqrt(np.mean(n.values[maska])))


btab = pd.DataFrame(
    {f"M>={m1}/{m2}": [implikowane_b(m1, m2, *d) for d in DEK] for m1, m2 in PARY},
    index=[f"{od}-{do}" for od, do in DEK])
btab.loc["rozstep"] = btab.max() - btab.min()
print("Implikowany wykladnik b po dekadach:")
print(btab.round(2).to_string())

ptab = pd.DataFrame({e: [nadwyzka(m, *z) for m in PROGI_P] for e, z in POLOWKI.items()},
                    index=PROGI_P)
print("\nZmiennosc rocznika w jednostkach podlogi Poissona (krotnosc):")
print(ptab.round(2).to_string())

# SKLADOWA X wydluzona z 4,00 na 5,80 (prosba Macka, 06.09).
fig, (a1, a2) = plt.subplots(1, 2, figsize=(2.5*5.80, 2.5*1.68),
                             gridspec_kw=dict(wspace=0.30))

srodki = [(od + do) / 2 for od, do in DEK]
for (m1, m2), c, dy in zip(PARY, [POMARANCZ, NIEBIESKI], [-16, 8]):
    kol_b = btab.loc[btab.index != "rozstep", f"M>={m1}/{m2}"]
    a1.plot(srodki, kol_b.values, color=c, lw=2.0, marker="o", ms=5)
    a1.annotate(f"M $\\geq$ {m1} / {m2}", (srodki[-1], kol_b.values[-1]),
                xytext=(-3, dy), textcoords="offset points", ha="right",
                color=c, fontsize=10 * CZCIONKA, fontweight="bold")
a1.axhline(1.0, color=SZARY, lw=1.2, ls=(0, (4, 3)), zorder=0)
a1.annotate("complete catalogue", (srodki[0], 1.0), xytext=(-2, 5),
            textcoords="offset points", color=TEKST2, fontsize=9 * CZCIONKA, va="bottom")
a1.set_ylabel("implied Gutenberg-Richter $b$")
a1.set_xlabel("decade")
a1.set_ylim(0.0, 1.35)
a1.set_title("$b$ must be constant in time", fontsize=10 * CZCIONKA, color=TEKST2,
             loc="left", pad=6)

xb = np.arange(len(PROGI_P))
sz = 0.36
for i, (e, c) in enumerate(zip(POLOWKI, [POMARANCZ, NIEBIESKI])):
    a2.bar(xb + (i - 0.5) * (sz + 0.02), ptab[e], width=sz, color=c, zorder=2,
           label=e)
a2.axhline(1.0, color=SZARY, lw=1.2, zorder=3)
a2.set_xticks(xb)
a2.set_xticklabels([f"{m}" for m in PROGI_P])
a2.set_xlabel("magnitude threshold")
a2.set_ylabel("year-to-year variation / Poisson floor")
a2.set_yscale("log")
a2.set_yticks([1, 2, 5, 10, 20])
a2.get_yaxis().set_major_formatter(mpl.ticker.FuncFormatter(lambda v, p: f"{v:g}$\\times$"))
a2.set_ylim(0.75, 30)
a2.grid(axis="x", visible=False)
a2.legend(loc="upper right", ncol=1, handlelength=1.2)
a2.set_title("how much the annual count really varies", fontsize=10 * CZCIONKA,
             color=TEKST2, loc="left", pad=6)
fig.tight_layout()
ZAPISANE.append(zapisz(fig, "instrument"))
```

```python
# RYSUNEK 6 (plansza 7) - TEZA POSTAWIONA WPROST, Z POKAZANIEM ODTRENDOWANIA.
#
# CO ZASTEPUJE I DLACZEGO. Poprzednia wersja miala dwa panele: rocznik po
# odtrendowaniu nad strumieniem CR. Maciek, 06.09: "nie ma przedstawionej
# nigdzie motywacji, aby cokolwiek odtrendowywac; wykres tez nie pokazuje,
# jak to odtrendowanie zostalo przeprowadzone; ten slajd jest do reworku".
# Oba zarzuty sa trafione - rysunek pokazywal WYNIK operacji, ktorej nigdzie
# nie bylo widac.
#
# NOWY UKLAD, dwa panele, wspolna os czasu:
#   GORNY  - surowy rocznik M >= 5,5 i dopasowany do niego gladki trend.
#            To jest cala operacja odtrendowania, pokazana wprost: odejmujemy
#            szara linie od niebieskiej.
#   DOLNY  - to, co zostaje, i strumien CR - OBA ZESTANDARYZOWANE (odjeta
#            srednia, podzielone przez odchylenie standardowe).
#
# DLACZEGO STANDARYZACJA, A NIE DWA PANELE JAK POPRZEDNIO. Poprzednio unikalem
# nakladania dwoch serii na jedna os, bo wymaga to wyboru skalowania, a widz
# oglada wtedy nasz wybor, nie dane. Standaryzacja tego problemu nie ma: to
# nie jest dobrany mnoznik, tylko sprowadzenie obu serii do zerowej sredniej i
# jednostkowej wariancji, czyli dokladnie tej postaci, w ktorej liczy sie
# wspolczynnik korelacji. Pytanie "czy te dwie krzywe chodza razem" jest wiec
# zadane wprost i widac je golym okiem.
#
# PO CO W OGOLE ODTRENDOWANIE (to musi wybrzmiec takze na planszy):
# obie serie maja wieloletni dryf. Korelacja policzona na seriach z dryfem
# mierzy przede wszystkim to, ze obie w ciagu 60 lat gdzies powedrowaly - a
# nie to, czy rocznik podaza za ~11-letnim cyklem. Odejmujemy wiec gladki
# trend z obu stron i pytamy o to, co zostaje.
PROG_GLOWNY = 5.5
STOPIEN = 3

n = licznik(PROG_GLOWNY)
trend = np.polyval(np.polyfit(LATA, n.values, STOPIEN), LATA)
reszta = pd.Series(n.values - trend, index=LATA)

mosc = pd.read_csv(f"{DATA_DIR}/mosc_data.csv")
mosc["datetime"] = pd.to_datetime(mosc["datetime"])
cr_rok = mosc.set_index("datetime")["value"].groupby(lambda t: t.year).mean().reindex(LATA)

wspolne = reszta.index[np.isfinite(reszta.values) & np.isfinite(cr_rok.values)]
r = float(np.corrcoef(reszta.loc[wspolne], cr_rok.loc[wspolne])[0, 1])
print(f"Korelacja reszty rocznika (M >= {PROG_GLOWNY}) ze strumieniem CR: r = {r:+.3f} "
      f"na {len(wspolne)} latach")
print(f"Rocznik M >= {PROG_GLOWNY}: {n.min():.0f}-{n.max():.0f} zdarzen/rok; "
      f"trend wielomianowy stopnia {STOPIEN} zmienia sie od {trend.min():.0f} "
      f"do {trend.max():.0f}")


def z(seria):
    v = np.asarray(seria, dtype=float)
    return (v - np.nanmean(v)) / np.nanstd(v)


fig, (a1, a2) = plt.subplots(2, 1, figsize=(2.5*4.00, 2.5*1.77), sharex=True,
                             gridspec_kw=dict(hspace=0.18))

a1.plot(n.index, n.values, color=NIEBIESKI, lw=2.0, zorder=2)
a1.plot(LATA, trend, color=SZARY, lw=2.0, ls=(0, (5, 3)), zorder=3)
a1.annotate(f"annual count, M $\\geq$ {PROG_GLOWNY}", (0.012, 0.94),
            xycoords="axes fraction", color=NIEBIESKI, fontsize=10 * CZCIONKA,
            fontweight="bold", va="top")
a1.annotate("smooth trend, subtracted", (0.012, 0.10), xycoords="axes fraction",
            color=TEKST2, fontsize=10 * CZCIONKA, fontweight="bold", va="bottom")
a1.set_ylabel("events per year")

a2.axhline(0, color=SZARY, lw=1.0, zorder=1)
a2.plot(LATA, z(cr_rok.values), color=POMARANCZ, lw=2.0, zorder=2)
a2.plot(LATA, z(reszta.values), color=NIEBIESKI, lw=2.0, zorder=3)
a2.annotate("what is left", (0.012, 0.94), xycoords="axes fraction",
            color=NIEBIESKI, fontsize=10 * CZCIONKA, fontweight="bold", va="top")
a2.annotate("cosmic-ray flux", (0.012, 0.06), xycoords="axes fraction",
            color=POMARANCZ, fontsize=10 * CZCIONKA, fontweight="bold", va="bottom")
a2.set_ylabel("standardised")
a2.set_xlabel("year")
a2.set_ylim(-3.2, 3.2)
# Swiadomie BEZ wspolczynnika korelacji na rysunku: plansza podaje liczby z
# 20260902d (trzy metody odtrendowania, po karze Bonferroniego), a ten rysunek
# uzywa jednej, wlasnej metody. Dwie rozne liczby obok siebie mylilyby.
a1.set_xlim(LATA[0] - 1, LATA[-1] + 1)
fig.tight_layout()
ZAPISANE.append(zapisz(fig, "correlation"))
```

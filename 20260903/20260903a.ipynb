```python
# 03.09.2026 - 20260903a.ipynb: SKAN PO LAGU Z ROZKLADEM ZEROWYM.
#
# JEDEN watek (konwencja repo: jeden watek = jeden notebook).
#
# PYTANIE: czy statystyka artykulu pokazuje sygnature PREKURSOROWA, o ktorej
# mowi sam artykul - maksimum przy dodatnim przesunieciu okolo +15 dni, spojne
# miedzy stacjami - czy jest na to przesuniecie slepa.
#
# DLACZEGO TO JEST PYTANIE, KTOREGO NAM BRAKUJE. Wszystko, co pokazalismy
# miedzy 26.08 a 02.09, ma postac "obserwowana struktura nie potrzebuje
# efektu": Fig.4 odtwarza sie na katalogu znajacym tylko roczna liczbe zdarzen
# (20260902b), wystarcza sama skladowa rosnaca (20260902c), preferowana faza
# cyklu jest wlasnoscia strony CR (20260902c), rocznik nie koreluje z CR po
# karze za proby (20260902d), a nasz lancuch wypada slabiej niz 500 z 500
# surogatow (20260902a). Ani jeden z tych rachunkow nie sprawdza WLASNEJ
# FALSYFIKOWALNEJ PRZEWIDYWALNI ARTYKULU: ze sygnal jest prekursorowy, czyli
# ze siedzi przy dodatnim lagu rzedu 15 dni. Homola prosil o skan po lagu
# wprost dwa razy (uwagi [o] i [ah]) i do dzis tego nie zrobilismy.
#
# DLACZEGO TEN TEST JEST CZYSTSZY NIZ WSZYSTKIE POPRZEDNIE. Caly spor
# ostatniego tygodnia dotyczy tego, ze wieloletni trend detekcyjny katalogu
# podnosi krzywa i produkuje minima. Ale trend dziala w skali DEKAD, a lag
# przesuwa kosze o DNI - artefakt trendowy nie ma zadnego powodu preferowac
# ktoregokolwiek lagu. Ksztalt krzywej po lagu jest wiec wolny od tego, co
# zatruwa wszystkie dotychczasowe liczby, i to niezaleznie od jej wysokosci.
#
# KIERUNEK LAGU - do sprawdzenia w komorce ze statystyka: kosze sejsmiczne sa
# przesuniete o +lag wzgledem koszy CR (krawedzie + LAG). Lag DODATNI znaczy
# wiec, ze trzesienia liczymy PO odpowiadajacym im koszu promieniowania
# kosmicznego, czyli CR wyprzedza EQ. To jest kierunek prekursorowy z artykulu.
# Artykul uzywa dt = 15 dni.
#
# DWIE ROZDZIELCZOSCI TEGO SAMEGO PYTANIA:
#
#  CZESC 1 - LAG LOKALNIE. W USTALONYCH t0 pieciu er (znalezionych na krzywych
#  z 26.08 tym samym kryterium co 31.08) liczymy -log10(PPDF) w funkcji lagu
#  od -60 do +60 dni co 1 dzien, dla katalogu prawdziwego i dla N = 500
#  katalogow surogatowych.
#  ZASTRZEZENIE, KTORE OBOWIAZUJE W CALEJ CZESCI 1: t0 pochodzi z prawdziwych
#  danych, wiec surogat jest oceniany w miejscu, w ktorym z definicji nie ma
#  powodu miec minimum. WYSOKOSCI krzywych NIE WOLNO tu porownywac - to byl
#  dokladnie ten blad, przed ktorym 02.09 uchronilo nas nieskalibrowanie
#  metody "stalego punktu". Porownujemy WYLACZNIE KSZTALT: kazda krzywa
#  (prawdziwa i surogatowa) jest przesuwana tak, zeby jej srednia po lagu byla
#  zerem. Pytanie brzmi "czy jest garb przy +15", a nie "czy jest wysoko".
#
#  CZESC 2 - LAG GLOBALNIE. Pelny skan po t0 powtorzony dla 15 wartosci lagu,
#  dla katalogu prawdziwego i 50 surogatow na stacje. Statystyka: lancuch T z
#  sekcji 4 (metoda "wolna", ta sama co 31.08 i 02.09) oraz liczba minimow.
#  Tu wysokosci porownywac WOLNO, bo t0 jest szukane od nowa dla kazdego lagu
#  i dla kazdego katalogu osobno - nikt nie dostaje daty z cudzych danych.
#
# REGULY ODCZYTU - USTALONE PRZED ZOBACZENIEM WYNIKU (02.09 kryterium w
# notebooku c okazalo sie za sztywne i trzeba bylo o tym pisac w dzienniku;
# tutaj definiujemy je z tolerancja od razu, ale NIE zmienimy ich po fakcie):
#
#   P. SYGNATURA PREKURSOROWA POTWIERDZONA, jesli LACZNIE:
#      - maksimum ksztaltu krzywej lezy w oknie +10...+20 dni u co najmniej
#        2 z 3 stacji (czesc 1, usrednienie po erach), ORAZ
#      - w tym oknie krzywa prawdziwa wychodzi ponad 95. centyl rozkladu
#        zerowego, ORAZ
#      - lancuch T (czesc 2) ma maksimum przy lagu dodatnim.
#      To bylby pierwszy od tygodnia wynik ZA artykulem i trzeba by go
#      zaraportowac tak samo uczciwie jak wszystkie przeciw.
#
#   N. LAG NIE NIESIE INFORMACJI, jesli krzywa prawdziwa miesci sie w pasmie
#      5-95 centyla rozkladu zerowego na calym zakresie lagow, albo jesli
#      maksima trzech stacji wypadaja w lagach niezgodnych ze soba (rozrzut
#      wiekszy niz 20 dni). Wtedy wybor dt = 15 dni nie ma oparcia we
#      wlasnych danych artykulu.
#
#   X. TEST NIEWAZNY W TEJ POSTACI, jesli sam rozklad zerowy silnie zalezy od
#      lagu (rozstep mediany surogatow po lagu wiekszy niz polowa rozstepu
#      krzywej prawdziwej). Znaczyloby to, ze lag zmienia cos w samej
#      konstrukcji koszy, a nie w zwiazku CR-EQ. Raportujemy to wprost,
#      zamiast czytac wynik.
#
# CZEGO TEN NOTEBOOK NIE ROZSTRZYGA: nie mierzy CZULOSCI statystyki, czyli nie
# odpowiada na pytanie, jak silny musialby byc efekt prekursorowy, zebysmy go
# tu zobaczyli. To wymaga testu z wstrzykiwaniem sygnalu (punkty otwarte
# 02.09). Wynik negatywny wolno wiec czytac jako "sygnatura, ktorej szuka
# artykul, nie wystaje ponad rozklad zerowy", a NIE jako "efektu nie ma".
#
# KOSZT [szacunek, traktowac jako rzad wielkosci - oszacowanie z 02.09 bylo
# zanizone pieciokrotnie]: czesc 1 okolo 2 min (501 katalogow x 15 krzywych
# po 121 punktow), czesc 2 okolo 7-15 min (153 zadania x 15 pelnych skanow).
import json
import os
import sys
import time

import numpy as np
import pandas as pd
from scipy.signal import find_peaks
from scipy.stats import binom

sys.path.insert(0, "..")
from mc_parallel import run_jobs_parallel

RESULTS_DIR = "../results"
DATA_DIR = "../data"
USGS_PATH = f"{DATA_DIR}/usgs_data/usgs_m4_1965_2025.csv"
MOSC_PATH = f"{DATA_DIR}/mosc_data.csv"
EXTENDED_DIR = f"{DATA_DIR}/csv_data_stations_extended"
FULL6H_DIR = f"{DATA_DIR}/csv_data_stations_full6h"

P_DAYS, D_DAYS, M_THRESHOLD = 3350, 5, 4.0
DT_ARTYKULU = 15                              # lag uzyty w artykule [dni]
BURNING_START = pd.Timestamp("2005-03-30")    # pierwszy rekord Augera (sekcja 3)
PROG_WYSOKOSCI = 5.0                          # prog find_peaks, kryterium z 31.08
STACJE = ["hrms", "aatb", "mosc"]             # dwie najsilniejsze wg 31.08 + stacja Fig.4

LAGI_GESTE = np.arange(-60, 61, 1)            # czesc 1: co 1 dzien
LAGI_RZADKIE = np.array([-60, -45, -30, -20, -15, -10, -5, 0,
                         5, 10, 15, 20, 30, 45, 60])   # czesc 2: pelne skany
N_SUR_CZESC1 = 500
N_SUR_CZESC2 = 50
ZIARNO_BAZOWE = 20260903


def load_mosc():
    df = pd.read_csv(MOSC_PATH)
    df["datetime"] = pd.to_datetime(df["datetime"])
    return df.set_index("datetime").sort_index()["value"]


def load_station(station):
    ext = os.path.join(EXTENDED_DIR, f"{station.lower()}_extended_6h.csv")
    full = os.path.join(FULL6H_DIR, f"{station.lower()}_full_6h.csv")
    df = pd.read_csv(ext if os.path.exists(ext) else full)
    df["datetime"] = pd.to_datetime(df["datetime"])
    return df.set_index("datetime").sort_index()["value"]


katalog = pd.read_csv(USGS_PATH, usecols=["time", "mag"])
katalog["time"] = pd.to_datetime(katalog["time"], utc=True).dt.tz_localize(None)
katalog = katalog[katalog["mag"] >= M_THRESHOLD].sort_values("time").reset_index(drop=True)
print(f"Katalog USGS M>={M_THRESHOLD}: {len(katalog)} zdarzen, "
      f"{katalog['time'].min().date()} .. {katalog['time'].max().date()}")
for st in STACJE:
    s = load_mosc() if st == "mosc" else load_station(st)
    print(f"CR {st:5s}: {len(s)} pomiarow, {s.index.min().date()} .. {s.index.max().date()}")
```

```python
# STATYSTYKA SPARAMETRYZOWANA LAGIEM. Kod przepisany z 20260902a/b (wersja
# szybka: searchsorted + add.reduceat, te same kosze co przepis referencyjny
# z 20260826d), z jedna zmiana: lag nie jest stala modulu, tylko argumentem.
# Nic poza tym sie nie zmienia - dla lag = 15 dni ta funkcja MUSI dawac to
# samo co krzywe zapisane 26.08, i to jest sprawdzane w komorce nizej.
_STEP_NS = np.int64(D_DAYS * 86400 * 10 ** 9)
_DZIEN_NS = np.int64(86400 * 10 ** 9)
_N_KOSZY = int(P_DAYS // D_DAYS)
_OFFSETY = _STEP_NS * np.arange(_N_KOSZY + 1, dtype=np.int64)


def przygotuj_szereg(s):
    t = s.index.values.astype("datetime64[ns]").astype(np.int64)
    v = s.to_numpy(dtype=float)
    ok = ~np.isnan(v)
    return t, np.append(np.where(ok, v, 0.0), 0.0), np.append(ok.astype(np.int64), 0)


def przygotuj_katalog(kat):
    return przygotuj_szereg(kat.set_index("time")["mag"].sort_index())


def _sumy_koszy(wartosci, idx):
    s = np.add.reduceat(wartosci, idx)[:-1]
    s[idx[1:] == idx[:-1]] = 0
    return s


def _znaki(cr_prep, eq_prep, t0_ns, lag_ns):
    krawedzie = t0_ns + _OFFSETY
    ct, cv, cn = cr_prep
    ci = np.searchsorted(ct, krawedzie, side="left")
    suma, liczba = _sumy_koszy(cv, ci), _sumy_koszy(cn, ci)
    with np.errstate(invalid="ignore", divide="ignore"):
        cr_vals = np.where(liczba > 0, suma / np.maximum(liczba, 1), np.nan)
    et, ev, _ = eq_prep
    # TU wchodzi lag: kosze sejsmiczne przesuniete o +lag wzgledem koszy CR.
    # lag > 0  ->  trzesienia liczone PO koszu CR  ->  CR wyprzedza EQ
    #              (kierunek prekursorowy, ten z artykulu).
    ei = np.searchsorted(et, krawedzie + lag_ns, side="left")
    sm_vals = _sumy_koszy(ev, ei)
    nCR_i, nCR_im1 = cr_vals[1:], cr_vals[:-1]
    dCR, Sm = nCR_i - nCR_im1, sm_vals[1:]
    A = Sm / np.nanmedian(Sm) - 1
    B = np.abs(dCR) / np.nanmedian(np.abs(dCR)) - 1
    valid = ((A != 0) & (B != 0) & (nCR_i > 0) & (nCR_im1 > 0) & (Sm > 0)
             & ~np.isnan(A) & ~np.isnan(B))
    c = (A * B)[valid]
    Np, Nm = int((c > 0).sum()), int((c < 0).sum())
    return Np, Np + Nm


def y_punkt(cr_prep, eq_prep, t0_ns, lag_dni):
    """-log10(PPDF) w JEDNYM punkcie t0, przy zadanym lagu."""
    Np, Nv = _znaki(cr_prep, eq_prep, np.int64(t0_ns), np.int64(lag_dni) * _DZIEN_NS)
    if Nv <= 0:
        return np.nan
    return float(-np.log10(binom.pmf(Np, Nv, 0.5)))


def y_krzywa(cr_prep, eq_prep, siatka_ns, lag_dni):
    """-log10(PPDF) na CALEJ siatce t0, przy zadanym lagu."""
    lag_ns = np.int64(lag_dni) * _DZIEN_NS
    Np = np.zeros(len(siatka_ns), dtype=np.int64)
    Nv = np.zeros(len(siatka_ns), dtype=np.int64)
    for i, t0_ns in enumerate(siatka_ns):
        Np[i], Nv[i] = _znaki(cr_prep, eq_prep, t0_ns, lag_ns)
    y = np.full(len(siatka_ns), np.nan)
    ok = Nv > 0
    y[ok] = -np.log10(binom.pmf(Np[ok], Nv[ok], 0.5))
    return y
```

```python
# SIATKI t0 I KRZYWE ODNIESIENIA. Siatki bierzemy WPROST z finetune_*.csv
# z 26.08 (tak samo jak 20260902a/b), zeby wszystko liczone dzisiaj bylo
# porownywalne punkt w punkt z tamtymi wynikami.
siatki, siatki_ns, finetune, cr_prep = {}, {}, {}, {}
for st in STACJE:
    ft = (pd.read_csv(f"{RESULTS_DIR}/finetune_{st}_P3350_d5_dt15_fullhistory.csv",
                      parse_dates=["t0"])
          .sort_values("t0").reset_index(drop=True))
    finetune[st] = ft
    siatki[st] = ft["t0"]
    siatki_ns[st] = ft["t0"].values.astype("datetime64[ns]").astype(np.int64)
    cr_prep[st] = przygotuj_szereg(load_mosc() if st == "mosc" else load_station(st))
    print(f"{st:5s}: siatka {len(ft)} punktow, "
          f"{ft['t0'].iloc[0].date()} .. {ft['t0'].iloc[-1].date()}")

KROK_DNI = (siatki[STACJE[0]].iloc[1] - siatki[STACJE[0]].iloc[0]) / pd.Timedelta(days=1)
DISTANCE = int(round(P_DAYS / KROK_DNI))
print(f"\nKrok siatki {KROK_DNI * 24:.0f} h; find_peaks: prog {PROG_WYSOKOSCI}, "
      f"distance {DISTANCE} probek (= {P_DAYS} dni)")

# KONTROLA 1 - bez niej nie wolno czytac reszty notebooka. Nasza funkcja przy
# lagu 15 dni (czyli przy dt artykulu) musi odtworzyc krzywa zapisana 26.08.
# Jesli nie odtworzy, to skanujemy po lagu inna statystyke niz ta, o ktora
# chodzi. Rozjazdy rzedu 0,1 sa dopuszczalne i znane (przyspieszona arytmetyka,
# opisana w 20260902a) - rozjazd systematyczny nie jest.
kat_prep_prawdziwy = przygotuj_katalog(katalog)
print("\nKontrola: nasza statystyka przy lag = 15 dni kontra krzywe z 26.08")
for st in STACJE:
    y_nasza = y_krzywa(cr_prep[st], kat_prep_prawdziwy, siatki_ns[st], DT_ARTYKULU)
    y_stara = -np.log10(finetune[st]["PPDF"].to_numpy())
    d = np.abs(y_nasza - y_stara)
    r = np.corrcoef(y_nasza[~np.isnan(d)], y_stara[~np.isnan(d)])[0, 1]
    print(f"  {st:5s} korelacja {r:.6f}, mediana |roznicy| {np.nanmedian(d):.5f}, "
          f"punktow z |roznica| > 0,2: {int(np.nansum(d > 0.2))} / {len(d)}")
```

```python
# ERY: t0, w ktorych liczymy krzywa po lagu (czesc 1). Znajdujemy je find_peaks
# wprost na krzywej z 26.08, tym samym kryterium co 31.08 i 02.09 (prog 5,0,
# nienakladajace sie okna 3350 dni). Nie przepisujemy dat recznie z dziennika -
# maja wyjsc z danych, zeby notebook byl samodzielny.
ery = {}
print("Ery (minima krzywej z 26.08) - to sa punkty, w ktorych skanujemy po lagu:")
for st in STACJE:
    y = -np.log10(finetune[st]["PPDF"].to_numpy())
    piki, _ = find_peaks(np.nan_to_num(y, nan=-1.0), height=PROG_WYSOKOSCI,
                         distance=DISTANCE)
    ery[st] = [(str(siatki[st].iloc[i].date()), int(siatki_ns[st][i]), float(y[i]))
               for i in piki]
    print(f"  {st:5s} ({len(piki)} er): " + ", ".join(
        f"{d} (glebokosc {h:.2f})" for d, _, h in ery[st]))

# Do wykresow potrzebna jest jeszcze etykieta ery wspolna miedzy stacjami -
# bierzemy rok t0.
for st in STACJE:
    ery[st] = [(d, ns, h, pd.Timestamp(d).year) for d, ns, h in ery[st]]
```

```python
# GENERATOR KATALOGU SUROGATOWEGO. Ten sam co 20260902a/b/c: dla kazdego roku
# DOKLADNIE tyle zdarzen, ile bylo naprawde, magnitudy losowane ze zwracaniem
# z tego samego roku, czasy losowe w obrebie roku (a scisle: w obrebie tej
# czesci roku, ktora katalog faktycznie pokrywa - poprawka z 02.09 dotyczaca
# niepelnego roku 2025).
#
# Ten surogat zachowuje CALA wieloletnia strukture katalogu (w tym hipotetyczna
# modulacje ~11-letnia) i niszczy wszystko szybsze niz rok - czyli dokladnie to,
# w czym mialby siedziec efekt prekursorowy z lagiem 15 dni. Do skanu po lagu
# jest to wlasciwy rozklad zerowy, w odroznieniu od pytan z 02.09.
_KAT_OD, _KAT_DO = katalog["time"].min(), katalog["time"].max()
_LATA = {int(r): g["mag"].to_numpy() for r, g in katalog.groupby(katalog["time"].dt.year)}
_GRANICE_LAT = {}
for _r in _LATA:
    _od = max(pd.Timestamp(year=_r, month=1, day=1), _KAT_OD)
    _do = min(pd.Timestamp(year=_r + 1, month=1, day=1), _KAT_DO)
    _GRANICE_LAT[_r] = (_od, (_do - _od).total_seconds())


def surogat_z_trendem(ziarno):
    rng = np.random.default_rng(ziarno)
    cz, mg = [], []
    for rok, magi in _LATA.items():
        poczatek, dlugosc = _GRANICE_LAT[rok]
        cz.append(poczatek + pd.to_timedelta(rng.uniform(0, dlugosc, len(magi)), unit="s"))
        mg.append(rng.choice(magi, size=len(magi), replace=True))
    s = pd.DataFrame(dict(time=np.concatenate(cz), mag=np.concatenate(mg)))
    s = s.sort_values("time").reset_index(drop=True)
    s["time"] = s["time"].dt.floor("s").astype("datetime64[us]")
    return s


# KONTROLA 2: surogat ma miec roczne liczby zdarzen zgodne CO DO JEDNEGO.
_probny = surogat_z_trendem(ZIARNO_BAZOWE)
_por = pd.DataFrame({"prawdziwy": katalog.groupby(katalog["time"].dt.year).size(),
                     "surogat": _probny.groupby(_probny["time"].dt.year).size()})
print(f"Surogat: {len(_probny)} zdarzen (prawdziwy {len(katalog)}), "
      f"najwieksza roznica rocznej liczby zdarzen "
      f"{int((_por['surogat'] - _por['prawdziwy']).abs().max())} (ma byc 0)")
print(f"Srednia magnituda: surogat {_probny['mag'].mean():.4f}, "
      f"prawdziwy {katalog['mag'].mean():.4f}")
del _probny, _por
```

```python
# CZESC 1 - LAG LOKALNIE. Jedno zadanie = jeden katalog (prawdziwy lub jeden
# surogat) przeliczony we wszystkich erach wszystkich stacji, po calej gestej
# siatce lagow. Katalog generujemy RAZ na zadanie i uzywamy do wszystkich
# stacji - dzieki temu porownanie miedzy stacjami jest sparowane.
SCIEZKA_C1 = f"{RESULTS_DIR}/lag_czesc1_krzywe.csv"
SCIEZKA_C1_KONFIG = f"{RESULTS_DIR}/lag_czesc1_config.json"
PRZELICZ_OD_NOWA = False


def konfiguracja_c1():
    return dict(STACJE=STACJE, P_DAYS=P_DAYS, D_DAYS=D_DAYS, M_THRESHOLD=M_THRESHOLD,
                PROG_WYSOKOSCI=PROG_WYSOKOSCI, N_SUR=N_SUR_CZESC1,
                ZIARNO_BAZOWE=ZIARNO_BAZOWE,
                LAGI=[int(LAGI_GESTE[0]), int(LAGI_GESTE[-1]), int(len(LAGI_GESTE))],
                ery={st: [d for d, _, _, _ in ery[st]] for st in STACJE},
                katalog_n=int(len(katalog)))


def zadanie_c1(ziarno):
    kat = katalog if ziarno < 0 else surogat_z_trendem(ziarno)
    eq = przygotuj_katalog(kat)
    wiersze = []
    for st in STACJE:
        for data, t0_ns, _, rok in ery[st]:
            y = np.array([y_punkt(cr_prep[st], eq, t0_ns, lag) for lag in LAGI_GESTE])
            wiersze.append(dict(typ="prawdziwy" if ziarno < 0 else "surogat",
                                ziarno=int(ziarno), stacja=st, era=rok, t0=data,
                                krzywa=";".join(f"{v:.5f}" for v in y)))
    return wiersze


zadania_c1 = [-1] + [ZIARNO_BAZOWE + i for i in range(N_SUR_CZESC1)]

wczytano_c1 = False
if not PRZELICZ_OD_NOWA and os.path.exists(SCIEZKA_C1) and os.path.exists(SCIEZKA_C1_KONFIG):
    if json.load(open(SCIEZKA_C1_KONFIG)) == json.loads(json.dumps(konfiguracja_c1())):
        c1 = pd.read_csv(SCIEZKA_C1)
        wczytano_c1 = len(c1) == len(zadania_c1) * sum(len(ery[s]) for s in STACJE)
        print(f"WCZYTANO gotowe krzywe z {SCIEZKA_C1} - nic nie liczono."
              if wczytano_c1 else "Zapisany plik nie ma tylu wierszy - liczymy od nowa.")
    else:
        print("Zapisana konfiguracja nie pasuje do biezacej - liczymy od nowa.")

if not wczytano_c1:
    print(f"Czesc 1: {len(zadania_c1)} katalogow x "
          f"{sum(len(ery[s]) for s in STACJE)} (stacja, era) x {len(LAGI_GESTE)} lagow.")
    start = time.time()
    c1 = run_jobs_parallel(zadanie_c1, zadania_c1, opis="katalogow", raport_co=50)
    print(f"Czas calkowity: {(time.time() - start) / 60:.1f} min")
    c1.to_csv(SCIEZKA_C1, index=False)
    json.dump(konfiguracja_c1(), open(SCIEZKA_C1_KONFIG, "w"), indent=1)
    print(f"Zapisano {SCIEZKA_C1}")

Y1 = np.array([[float(x) for x in s.split(";")] for s in c1["krzywa"]])
print(f"\nMacierz krzywych: {Y1.shape} (wierszy x lagow)")
```

```python
# CZESC 1 - LICZBY. Ksztalt, nie wysokosc: kazda krzywa przesuwamy tak, zeby
# jej wlasna srednia po lagu byla zerem (patrz zastrzezenie w komorce 0).
Z1 = Y1 - np.nanmean(Y1, axis=1, keepdims=True)
IDX_PRAWDZIWY = (c1["typ"] == "prawdziwy").to_numpy()
LAG15 = int(np.where(LAGI_GESTE == DT_ARTYKULU)[0][0])

def _rozstep(x):
    return float(np.nanmax(x) - np.nanmin(x))


podsum_c1, krzywe_c1 = [], {}
for st in STACJE:
    m_st = (c1["stacja"] == st).to_numpy()
    # usrednienie po erach w obrebie jednej realizacji
    praw = np.nanmean(Z1[m_st & IDX_PRAWDZIWY], axis=0)
    ziarna = sorted(c1.loc[m_st & ~IDX_PRAWDZIWY, "ziarno"].unique())
    zi = c1["ziarno"].to_numpy()
    sur = np.stack([np.nanmean(Z1[m_st & (zi == z)], axis=0) for z in ziarna])
    p05, p50, p95 = np.percentile(sur, [5, 50, 95], axis=0)
    arg_praw = int(LAGI_GESTE[np.nanargmax(praw)])
    arg_sur = LAGI_GESTE[np.nanargmax(sur, axis=1)]
    # p-wartosc dla wartosci przy lagu artykulu, estymator Davisona-Hinkleya
    p_emp15 = (1 + int((sur[:, LAG15] >= praw[LAG15]).sum())) / (1 + len(sur))
    podsum_c1.append(dict(stacja=st, arg_max_prawdziwy=arg_praw,
                          z_przy_lag15=praw[LAG15], p95_przy_lag15=p95[LAG15],
                          p_emp_lag15=p_emp15,
                          rozstep_prawdziwej=_rozstep(praw),
                          rozstep_mediany_surogatow=_rozstep(p50),
                          frakcja_surogatow_argmax_10_20=float(
                              np.mean((arg_sur >= 10) & (arg_sur <= 20))),
                          ponad_p95=bool(praw[LAG15] > p95[LAG15])))
    krzywe_c1[st] = (praw, p05, p50, p95, arg_sur)

podsum_c1 = pd.DataFrame(podsum_c1)
pd.set_option("display.width", 200)
print("CZESC 1 - KSZTALT KRZYWEJ PO LAGU (usrednione po erach danej stacji)")
print(podsum_c1.to_string(index=False, float_format=lambda v: f"{v:.4f}"))
print()
print("Jak to czytac:")
print(" - arg_max_prawdziwy: lag [dni], przy ktorym krzywa prawdziwa jest najwyzsza.")
print(f"   Artykul przewiduje okolo +{DT_ARTYKULU}; regula P wymaga okna +10...+20.")
print(" - p_emp_lag15: jaka czesc surogatow ma przy lagu 15 ksztalt co najmniej")
print("   tak wysoki jak katalog prawdziwy.")
print(" - rozstep_mediany_surogatow kontra rozstep_prawdziwej: regula X (test")
print("   niewazny) wlacza sie, gdy ten pierwszy przekracza polowe drugiego.")
```

```python
# CZESC 2 - LAG GLOBALNIE. Pelny skan po t0 powtorzony dla kazdego lagu z
# rzadkiej siatki. Statystyka: lancuch kary z sekcji 4 (metoda "wolna", wzor
# identyczny jak w 20260831b / 20260901b / 20260902a) oraz liczba minimow.
#
# Tu wysokosc porownywac WOLNO: dla kazdego lagu i kazdego katalogu minima sa
# szukane od nowa na calej siatce, wiec zaden katalog nie dostaje daty
# wyliczonej z cudzych danych.
def lancuch_wolny(y, siatka, wariant, data_start, data_end):
    piki, _ = find_peaks(np.nan_to_num(y, nan=-1.0), height=PROG_WYSOKOSCI,
                         distance=DISTANCE)
    t0_pikow = siatka.iloc[piki].reset_index(drop=True)
    wys_pikow = y[piki]
    if wariant == "a_z_wykluczeniem":
        maska = (t0_pikow < BURNING_START).to_numpy()
        granica = BURNING_START
    else:
        maska = np.ones(len(t0_pikow), dtype=bool)
        granica = data_end
    t0_uzyte, wys_uzyte = t0_pikow[maska].reset_index(drop=True), wys_pikow[maska]
    prev, T, przerwany = data_start, 0.0, False
    for t0_i, h in zip(t0_uzyte, wys_uzyte):
        n = (granica - prev).days / D_DAYS
        if n <= 0:
            przerwany = True
            break
        T += h - np.log10(n)
        prev = t0_i
    return dict(T=float(T), n_minimow=int(len(t0_uzyte)), przerwany=bool(przerwany),
                najglebsze=float(wys_uzyte.max()) if len(wys_uzyte) else np.nan)


# KONTROLA 3: lancuch przy lagu 15 dni, policzony na krzywych z 26.08, musi
# odtworzyc liczby z dziennika 31.08 - inaczej skanujemy po lagu inny lancuch.
ODNIESIENIE = {("hrms", "a_z_wykluczeniem"): 33.684, ("hrms", "b_bez_wykluczenia"): 45.707,
               ("aatb", "a_z_wykluczeniem"): 17.671, ("aatb", "b_bez_wykluczenia"): 22.179,
               ("mosc", "a_z_wykluczeniem"): 16.737, ("mosc", "b_bez_wykluczenia"): 21.867}
print("Kontrola: lancuch odtworzony z krzywych 26.08 kontra dziennik 31.08")
for st in STACJE:
    y = -np.log10(finetune[st]["PPDF"].to_numpy())
    for wariant in ("a_z_wykluczeniem", "b_bez_wykluczenia"):
        r = lancuch_wolny(y, siatki[st], wariant, siatki[st].iloc[0], siatki[st].iloc[-1])
        ref = ODNIESIENIE.get((st, wariant))
        zgoda = "OK" if ref is not None and abs(r["T"] - ref) < 0.01 else ">>> ROZJAZD"
        print(f"  {st:5s} {wariant:18s} minimow={r['n_minimow']}  T={r['T']:.3f}  "
              f"(31.08: {ref:.3f})  {zgoda}")
```

```python
# CZESC 2 - RACHUNEK. Jedno zadanie = jedna (stacja, katalog); w srodku petla
# po lagach, zeby katalog surogatowy byl generowany raz na zadanie.
SCIEZKA_C2 = f"{RESULTS_DIR}/lag_czesc2_lancuch.csv"
SCIEZKA_C2_KONFIG = f"{RESULTS_DIR}/lag_czesc2_config.json"


def konfiguracja_c2():
    return dict(STACJE=STACJE, P_DAYS=P_DAYS, D_DAYS=D_DAYS, M_THRESHOLD=M_THRESHOLD,
                PROG_WYSOKOSCI=PROG_WYSOKOSCI, N_SUR=N_SUR_CZESC2,
                ZIARNO_BAZOWE=ZIARNO_BAZOWE, LAGI=[int(x) for x in LAGI_RZADKIE],
                BURNING_START=str(BURNING_START), katalog_n=int(len(katalog)))


def zadanie_c2(job):
    st, ziarno = job
    kat = katalog if ziarno < 0 else surogat_z_trendem(ziarno)
    eq = przygotuj_katalog(kat)
    start, koniec = siatki[st].iloc[0], siatki[st].iloc[-1]
    wiersze = []
    for lag in LAGI_RZADKIE:
        y = y_krzywa(cr_prep[st], eq, siatki_ns[st], lag)
        for wariant in ("a_z_wykluczeniem", "b_bez_wykluczenia"):
            r = lancuch_wolny(y, siatki[st], wariant, start, koniec)
            r.update(typ="prawdziwy" if ziarno < 0 else "surogat", ziarno=int(ziarno),
                     stacja=st, lag=int(lag), wariant=wariant,
                     max_krzywej=float(np.nanmax(y)))
            wiersze.append(r)
    return wiersze


zadania_c2 = ([(st, -1) for st in STACJE]
              + [(st, ZIARNO_BAZOWE + i) for st in STACJE for i in range(N_SUR_CZESC2)])

wczytano_c2 = False
if not PRZELICZ_OD_NOWA and os.path.exists(SCIEZKA_C2) and os.path.exists(SCIEZKA_C2_KONFIG):
    if json.load(open(SCIEZKA_C2_KONFIG)) == json.loads(json.dumps(konfiguracja_c2())):
        c2 = pd.read_csv(SCIEZKA_C2)
        wczytano_c2 = len(c2) == len(zadania_c2) * len(LAGI_RZADKIE) * 2
        print(f"WCZYTANO gotowe wyniki z {SCIEZKA_C2} - nic nie liczono."
              if wczytano_c2 else "Zapisany plik nie ma tylu wierszy - liczymy od nowa.")
    else:
        print("Zapisana konfiguracja nie pasuje do biezacej - liczymy od nowa.")

if not wczytano_c2:
    print(f"Czesc 2: {len(zadania_c2)} zadan x {len(LAGI_RZADKIE)} lagow x "
          f"pelny skan po t0.")
    start = time.time()
    c2 = run_jobs_parallel(zadanie_c2, zadania_c2, opis="zadan", raport_co=10)
    print(f"Czas calkowity: {(time.time() - start) / 60:.1f} min")
    c2.to_csv(SCIEZKA_C2, index=False)
    json.dump(konfiguracja_c2(), open(SCIEZKA_C2_KONFIG, "w"), indent=1)
    print(f"Zapisano {SCIEZKA_C2}")
```

```python
# CZESC 2 - LICZBY. Wariant a_z_wykluczeniem (ten, na ktorym stoja liczby
# z 31.08); wariant b jest w pliku i wchodzi na wykres jako kontrola.
WARIANT = "a_z_wykluczeniem"
c2w = c2[c2["wariant"] == WARIANT]
podsum_c2, krzywe_c2 = [], {}
for st in STACJE:
    g = c2w[c2w["stacja"] == st]
    praw = g[g["typ"] == "prawdziwy"].set_index("lag")["T"].sort_index()
    sur = (g[g["typ"] == "surogat"].pivot_table(index="ziarno", columns="lag", values="T")
           .reindex(columns=sorted(praw.index)))
    p05, p50, p95 = (np.percentile(sur.to_numpy(), q, axis=0) for q in (5, 50, 95))
    p_emp = {lag: (1 + int((sur[lag] >= praw[lag]).sum())) / (1 + len(sur))
             for lag in praw.index}
    podsum_c2.append(dict(
        stacja=st,
        lag_max_prawdziwy=int(praw.idxmax()), T_max=float(praw.max()),
        T_przy_lag15=float(praw[DT_ARTYKULU]),
        T_bez_lagu=float(praw[0]),
        roznica_15_minus_0=float(praw[DT_ARTYKULU] - praw[0]),
        rozstep_T_prawdziwej=float(praw.max() - praw.min()),
        rozstep_mediany_surogatow=float(p50.max() - p50.min()),
        p_emp_lag15=p_emp[DT_ARTYKULU],
        lag_max_mediany_surogatow=int(sorted(praw.index)[int(np.argmax(p50))])))
    krzywe_c2[st] = (praw, p05, p50, p95)

podsum_c2 = pd.DataFrame(podsum_c2)
print(f"CZESC 2 - LANCUCH T W FUNKCJI LAGU (wariant {WARIANT})")
print(podsum_c2.to_string(index=False, float_format=lambda v: f"{v:.3f}"))
print()
print("Jak to czytac:")
print(" - roznica_15_minus_0: ile lancuch ZYSKUJE na tym, ze artykul przesuwa")
print("   serie sejsmiczna o 15 dni. Jesli to jest bliskie zeru albo ujemne,")
print("   przesuniecie prekursorowe nie robi dla wyniku artykulu nic.")
print(" - lag_max_mediany_surogatow: gdzie maksimum ma sam rozklad zerowy.")
print("   Jesli wypada tam samo co prawdziwy, to nie jest sygnal, tylko")
print("   wlasnosc konstrukcji koszy (regula X).")
```

```python
# CZESC 2, WYKRES 4 (jedno kodowanie: wysokosc krzywej). Lancuch T w funkcji
# lagu, panel na stacje, z pasmem rozkladu zerowego. Rozklad zerowy jest tu
# ZASADNY takze co do wysokosci - por. komorka z definicja czesci 2.
fig, osie = plt.subplots(len(STACJE), 1, figsize=(11, 3.2 * len(STACJE)), sharex=True)
for ax, st in zip(np.atleast_1d(osie), STACJE):
    praw, p05, p50, p95 = krzywe_c2[st]
    lagi = np.array(sorted(praw.index))
    ax.fill_between(lagi, p05, p95, color="#55A868", alpha=0.25,
                    label="rozkład zerowy (5–95 centyl)")
    ax.plot(lagi, p50, color="#55A868", lw=1.2, ls="--", label="mediana rozkładu zerowego")
    ax.plot(lagi, praw.to_numpy(), "o-", color="#4C72B0", lw=1.8, ms=4,
            label="katalog prawdziwy")
    ax.axvline(DT_ARTYKULU, color="#C44E52", lw=1.4,
               label=f"lag użyty w artykule (+{DT_ARTYKULU} dni)")
    ax.axvline(0, color="gray", lw=0.8, ls=":")
    ax.set_ylabel("T = −log₁₀(iloczyn łańcucha)")
    ax.set_title(f"stacja {st.upper()}")
np.atleast_1d(osie)[0].legend(loc="best", fontsize=8)
np.atleast_1d(osie)[-1].set_xlabel("przesunięcie serii sejsmicznej względem CR [dni]\n"
                                   "(dodatnie = promieniowanie kosmiczne wyprzedza trzęsienia)")
fig.suptitle("Istotność całego łańcucha z sekcji 4 w funkcji przesunięcia prekursorowego")
fig.tight_layout()
fig.savefig(f"{RESULTS_DIR}/lag_4_lancuch_po_lagu.png", dpi=150)
print(f"Zapisano {RESULTS_DIR}/lag_4_lancuch_po_lagu.png")
```

```python
# CZESC 2, WYKRES 5 (jedno kodowanie: wysokosc slupka). Liczba minimow powyzej
# progu w funkcji lagu. Osobny wykres, a nie druga seria na poprzednim: to jest
# inna wielkosc (licznik, nie istotnosc) i mieszanie ich na jednej osi
# zaciemnia oba.
fig, osie = plt.subplots(1, len(STACJE), figsize=(4.2 * len(STACJE), 3.6), sharey=True)
for ax, st in zip(np.atleast_1d(osie), STACJE):
    g = c2w[c2w["stacja"] == st]
    praw = g[g["typ"] == "prawdziwy"].set_index("lag")["n_minimow"].sort_index()
    sur = g[g["typ"] == "surogat"].groupby("lag")["n_minimow"].mean().sort_index()
    ax.plot(praw.index, praw.to_numpy(), "o-", color="#4C72B0", label="katalog prawdziwy")
    ax.plot(sur.index, sur.to_numpy(), "s--", color="#55A868", ms=4,
            label="średnia z surogatów")
    ax.axvline(DT_ARTYKULU, color="#C44E52", lw=1.2)
    ax.set_title(f"{st.upper()}")
    ax.set_xlabel("lag [dni]")
np.atleast_1d(osie)[0].set_ylabel("liczba minimów powyżej progu")
np.atleast_1d(osie)[0].legend(fontsize=8)
fig.suptitle("Liczba minimów w funkcji przesunięcia prekursorowego")
fig.tight_layout()
fig.savefig(f"{RESULTS_DIR}/lag_5_liczba_minimow_po_lagu.png", dpi=150)
print(f"Zapisano {RESULTS_DIR}/lag_5_liczba_minimow_po_lagu.png")
```
